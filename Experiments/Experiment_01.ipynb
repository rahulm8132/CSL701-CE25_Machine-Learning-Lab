Task 9: Fill missing Age values with the median.

```python
df["Age"].median()
df["Age"] = df["Age"].fillna(df["Age"].median())

print("Median Age:", df["Age"].median())
print("Missing Age values after filling:", df["Age"].isnull().sum())
```

Task 13: Plot histogram of Age.

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

Task 14: Plot bar chart of Survival Count.

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Survived", data=df)
plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Count")
plt.show()
```

Task 15: Plot scatter graph of Age vs Fare.

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Age"], df["Fare"])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```
